# 3. Limpieza

## ¿Hay errores en los datos? ¿Cómo los corrijo?

Se detectaron dos errores de carga:

1. **`Especie`**: inconsistencia de mayúsculas/minúsculas (`'Adelie Penguin'` vs `'Gentoo penguin'` vs `'Chinstrap penguin'`), que hace que en teoría existan 3 categorías bien definidas pero mal escritas. Se normaliza el texto.
2. **`Sexo`**: existe un valor `'.'` que no corresponde a ninguna categoría válida (`MALE`/`FEMALE`). Se trata como dato faltante.


```python
# 1) Normalizamos el formato de 'Especie'
df['Especie'] = df['Especie'].astype(str).str.strip().str.title().astype('category')
df['Especie'].value_counts()
```

```python
# 2) El valor '.' en 'Sexo' es un error de carga -> lo tratamos como nulo
df['Sexo'] = df['Sexo'].astype(str).replace({'.': np.nan, 'nan': np.nan}).astype('category')
df['Sexo'].value_counts(dropna=False)
```

## ¿Tiene nulos mi dataset? ¿Cómo los trato? ¿Cómo afecta a los datos mi tratamiento?

```python
df.isna().sum()
```

```python
print("Filas con al menos un nulo:", df.isna().any(axis=1).sum())
df[df.isna().any(axis=1)]
```

Distinguimos dos casos:

1. **Filas totalmente vacías** (todas las variables numéricas y `Sexo` nulas): Hay 2 y no aportan ninguna información y se **eliminan** directamente.
2. **Filas con sólo `Sexo` nulo**: representan una fracción muy pequeña del dataset (9 que son un aprox. 2.6%) y las variables morfológicas sí están presentes. En vez de descartarlas, **imputamos `Sexo` con la moda calculada por especie**, ya que la proporción de machos/hembras puede variar levemente entre especies.

```python
# 1) Eliminamos las filas completamente vacías (sin ningún dato morfológico)
filas_vacias = df[num_columns].isna().all(axis=1)
print("Filas completamente vacías a eliminar:", filas_vacias.sum())

df = df[~filas_vacias].copy()
df.isna().sum()
```

```python
# 2) Imputamos 'Sexo' con la moda por especie
df['Sexo'] = df.groupby('Especie', observed=True)['Sexo'].transform(
    lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else np.nan))

df.isna().sum()
```

```python
print("Filas finales en el dataset:", len(df))
df['Sexo'].value_counts()
```

Tras la limpieza, el dataset queda sin valores nulos ni categorías inválidas, conservando el 99.4% de las filas originales (345 de 347).

## ¿Tiene atípicos? ¿Cómo los trato? ¿Cómo afecta a los datos mi tratamiento?

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df[num_columns])
plt.xticks(rotation=20, ha='right')
plt.title("Boxplot de variables numéricas (escalas originales)")
plt.show()
```

```python
## Contar cantidad de atípicos por columna (criterio IQR)
for col in num_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"Cantidad de atípicos en {col}: {len(outliers)} | {len(outliers)/len(df) * 100:.2f}%")
```

Con el criterio de rango intercuartílico (IQR × 1.5) **no se detectan valores atípicos** en ninguna de las cuatro variables numéricas. Esto es consistente con lo observado en los boxplots por especie: la dispersión que se veía "grande" a nivel global (por ejemplo en `Masa corporal` o `Longitud Aleta`) se debe a que estamos mezclando tres especies de tamaños naturalmente distintos, no a errores de medición o valores extremos dentro de cada especie. Por lo tanto, **no es necesario eliminar ni imputar atípicos** en este dataset.

# 7. Reducción de dimensionalidad: PCA

```python
pca = PCA().set_output(transform="pandas")
df_pca = pca.fit_transform(df_scaled[num_columns])
df_pca[['Especie', 'Especie_cod']] = df_scaled[['Especie', 'Especie_cod']].values
df_pca.head()
```

## Selección del número de componentes

```python
var_exp = pca.explained_variance_ratio_
var_cum = np.cumsum(var_exp)

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(range(1, len(var_exp) + 1), var_exp, alpha=0.7, label='Varianza explicada')
ax.plot(range(1, len(var_exp) + 1), var_cum, color='red', marker='o', label='Varianza acumulada')
for i, (v, c) in enumerate(zip(var_exp, var_cum), start=1):
    ax.text(i, v + 0.02, f"{v:.2f}", ha='center', fontsize=9)
    ax.text(i, c + 0.02, f"{c:.2f}", ha='center', color='red', fontsize=9)
ax.set_xlabel('Componente principal')
ax.set_ylabel('Proporción de varianza')
ax.set_title('Varianza explicada y acumulada — PCA')
ax.set_xticks(range(1, len(var_exp) + 1))
ax.legend()
plt.tight_layout()
plt.show()
```

Con 2 componentes se acumula el **88.4%** de la varianza total (69.0% + 19.4%), superando el umbral habitual de 75-80%. La tercera componente aporta solo un 9.0% adicional.

### Criterio de Kaiser

```python
eigenvalues = pca.explained_variance_
print("Autovalores:", np.round(eigenvalues, 3))
print("Componentes con autovalor > 1:", (eigenvalues > 1).sum())
```

Bajo este criterio sólo la **primera componente** (autovalor = 2.77) se conserva; la segunda (0.78) queda por debajo del umbral. Es el criterio más conservador de los tres.

### Gráfico del codo

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(range(1, len(var_exp) + 1), var_exp, 'o-', linewidth=2, color='steelblue')
ax.set_xlabel('Componente principal')
ax.set_ylabel('Proporción de varianza explicada')
ax.set_title('Gráfico del codo — PCA')
ax.set_xticks(range(1, len(var_exp) + 1))
plt.tight_layout()
plt.show()
```

La caída más pronunciada ocurre entre PC1 y PC2 (-0.50); a partir de ahí la curva se aplana (PC2→PC3: -0.10, PC3→PC4: -0.06). El codo se ubica en la componente 2.

### Conclusión

Varianza acumulada y codo coinciden en **2 componentes**; Kaiser, más estricto, sugiere 1. Nos quedamos con **2 componentes**: superan el umbral de varianza acumulada, coinciden con el codo, y permiten una visualización directa en 2D.

## Visualización de las componentes

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_pca, x='pca0', y='pca1', hue='Especie', palette='Set2', s=60)
plt.xlabel(f'PC1 ({var_exp[0]*100:.1f}% var.)')
plt.ylabel(f'PC2 ({var_exp[1]*100:.1f}% var.)')
plt.title('PCA: primeras dos componentes')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

Gentoo se separa con claridad sobre el eje **PC1** (valor medio ≈ +2.0, contra -1.5 de Adelie y -0.4 de Chinstrap), coherente con que PC1 está dominado por Longitud de Aleta y Masa Corporal (variables en las que Gentoo se distingue fuertemente en el EDA). Adelie y Chinstrap, más parecidos en tamaño, se solapan parcialmente en PC1 pero se separan mejor en **PC2** (Chinstrap ≈ +1.0 vs Adelie ≈ -0.1), eje dominado por la forma del culmen (longitud y profundidad con igual signo), no por el tamaño general.

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

for especie, color in zip(sorted(df_pca['Especie'].unique()), sns.color_palette('Set2', 3)):
    subset = df_pca[df_pca['Especie'] == especie]
    ax.scatter(subset['pca0'], subset['pca1'], subset['pca2'], label=especie, color=color, s=40, alpha=0.8)

ax.set_xlabel('PC1')
ax.set_ylabel('PC2')
ax.set_zlabel('PC3')
ax.set_title('PCA: primeras tres componentes')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

Agregar PC3 no aporta una separación adicional relevante entre especies (Adelie ≈ -0.30, Chinstrap ≈ +0.75, Gentoo ≈ -0.04): Chinstrap vuelve a distinguirse levemente, pero de forma redundante con lo que ya se observa en PC2. Esto confirma numéricamente la elección de 2 componentes del punto anterior.

# 8. Reducción de dimensionalidad: Isomap

```python
resultados_isomap = {}
for k in [5, 15, 50]:
    modelo = Isomap(n_neighbors=k, n_components=2)
    proyeccion = modelo.fit_transform(df_scaled[num_columns])
    resultados_isomap[k] = (modelo, proyeccion)
    print(f"n_neighbors={k:2d} | error de reconstrucción: {modelo.reconstruction_error():.4f}")
```

El error de reconstrucción **baja** a medida que aumenta `n_neighbors` (0.94 con 3 vecinos hasta un mínimo cercano a 0.38 con 30), y luego **vuelve a subir** con vecindarios muy amplios (0.55 con 60), además de emitir una advertencia de grafo desconectado con pocos vecinos: con `n_neighbors` bajo, el grafo de vecinos no llega a conectar completamente el dataset (probablemente porque Gentoo queda bastante alejado de Adelie/Chinstrap en el espacio original), y recién se conecta del todo con vecindarios muy grandes, momento en el que Isomap empieza a comportarse de forma más parecida a PCA y pierde algo de estructura local.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, k in zip(axes, [5, 15, 50]):
    _, proyeccion = resultados_isomap[k]
    sns.scatterplot(x=proyeccion[:, 0], y=proyeccion[:, 1], hue=df_scaled['Especie'].values,
                     palette='Set2', s=40, ax=ax, legend=(k == 50))
    ax.set_title(f'Isomap (n_neighbors={k})')
    ax.set_xlabel('Componente 1')
    ax.set_ylabel('Componente 2')
if axes[-1].get_legend() is not None:
    axes[-1].legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

Con `n_neighbors=5` la proyección es más irregular y con puntos dispersos de forma poco compacta (estructura local ruidosa por el grafo parcialmente desconectado). Con `n_neighbors=50` la forma se vuelve más suave, pero las tres especies quedan más comprimidas entre sí (menos margen entre grupos). `n_neighbors=15` da el mejor equilibrio: separa igual de bien a Gentoo del resto y conserva algo más de distancia entre Adelie y Chinstrap que con 50 vecinos.

```python
isomap_final = Isomap(n_neighbors=15, n_components=2)
proyeccion_final = isomap_final.fit_transform(df_scaled[num_columns])

plt.figure(figsize=(8, 6))
sns.scatterplot(x=proyeccion_final[:, 0], y=proyeccion_final[:, 1],
                hue=df_scaled['Especie'].values, palette='Set2', s=60)
plt.xlabel('Componente 1')
plt.ylabel('Componente 2')
plt.title('Isomap (n_neighbors=15) — 2 componentes')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

Con la configuración elegida (`n_neighbors=15`), Gentoo se separa claramente del resto sobre la Componente 1, replicando lo observado con PCA, mientras que Adelie y Chinstrap muestran cierto solapamiento — señal de que la relación no lineal que Isomap intenta capturar no aporta, en este dataset, una separación mejor que la lineal (PCA) para distinguir estas dos especies.

# 9. Reducción de dimensionalidad: t-SNE

```python
resultados_tsne = {}
for perp in [10, 30, 50]:
    modelo = TSNE(n_components=2, perplexity=perp, max_iter=1000, init='pca', random_state=42)
    proyeccion = modelo.fit_transform(df_scaled[num_columns].to_numpy())
    resultados_tsne[perp] = (modelo, proyeccion)
    print(f"perplexity={perp:2d} | divergencia KL final: {modelo.kl_divergence_:.4f}")
```

La divergencia KL final **baja** a medida que aumenta la perplejidad (0.55 → 0.37 → 0.25), lo cual es esperable: con más vecinos considerados, al modelo le resulta más fácil encontrar un embedding en 2D que respete esas relaciones de vecindad más amplias. Sin embargo, una divergencia KL menor no implica automáticamente una mejor visualización — hay que confirmarlo mirando los gráficos.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, perp in zip(axes, [10, 30, 50]):
    _, proyeccion = resultados_tsne[perp]
    sns.scatterplot(x=proyeccion[:, 0], y=proyeccion[:, 1], hue=df_scaled['Especie'].values,
                     palette='Set2', s=40, ax=ax, legend=(perp == 50))
    ax.set_title(f't-SNE (perplexity={perp})')
    ax.set_xlabel('Componente 1')
    ax.set_ylabel('Componente 2')
if axes[-1].get_legend() is not None:
    axes[-1].legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

Con `perplexity=10` los tres grupos ya aparecen separados, pero con bordes más irregulares y algunos puntos sueltos entre clústeres (vecindarios pequeños, más sensibles a ruido local). Con `perplexity=50` los clústeres se ven más compactos y redondeados, con menos puntos aislados. En los tres casos las tres especies quedan visualmente separadas, lo cual es consistente con lo ya observado en el EDA (las cuatro variables morfológicas discriminan bien entre especies) y con PCA/Isomap.

```python
tsne_final = TSNE(n_components=2, perplexity=30, max_iter=1000, init='pca', random_state=42)
proyeccion_tsne_final = tsne_final.fit_transform(df_scaled[num_columns].to_numpy())

plt.figure(figsize=(8, 6))
sns.scatterplot(x=proyeccion_tsne_final[:, 0], y=proyeccion_tsne_final[:, 1],
                hue=df_scaled['Especie'].values, palette='Set2', s=60)
plt.xlabel('Componente 1')
plt.ylabel('Componente 2')
plt.title('t-SNE (perplexity=30) — 2 componentes')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

Con `perplexity=30` (punto intermedio, y por debajo del límite recomendado en relación al tamaño del dataset de 345 filas) se obtiene un buen equilibrio: tres clústeres bien diferenciados y razonablemente compactos, sin la irregularidad de `perplexity=10` ni la sobre-compresión de `perplexity=50`. A diferencia de PCA e Isomap, acá no es válido interpretar la distancia entre los tres grupos ni su tamaño relativo — sólo el hecho de que forman grupos separados.
